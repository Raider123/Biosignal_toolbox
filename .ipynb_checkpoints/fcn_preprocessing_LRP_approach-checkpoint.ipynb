# Functions

```python
def GetOnsetsAndMarker(raw, MarkerNumber): 
    #extracting marked events 
    events, event_dict = mne.events_from_annotations(raw)
    print("Events:", events)
    #print(events)

    # Marker before cropping 
    MarkerIndizesOld = events[:,0]
    MarkerNumbersOld = events[:,2]

    StartMarker = np.where(MarkerNumbersOld == 8)[0][0]
    StartMarkerIndex = MarkerIndizesOld[StartMarker]
    StopMarker = np.where(MarkerNumbersOld == 9)[0][0]
    StopMarkerIndex = MarkerIndizesOld[StopMarker]

    #Time points equal to Marker (in seconds)
    StartTimeScenario = raw.times[StartMarkerIndex]
    StopTimeScenario = raw.times[StopMarkerIndex]


    # Extract S100 (Movement Onset,leaving plate)
    S100Indizes = np.where(MarkerNumbersNew == Markernumber)[0]
    S100Marker = MarkerIndizesNew[S100Indizes]
    # use leaving plate (S100) as movement onset marker 
    OnsetMarkerIndex = np.copy(S100Marker) 
    
    return OnsetMarkerIndex

def RereferencingEpoching(raw, OnsetNumber, ErrorNumber,ChannelList,RerefChannels): 
    # rereferencing 
    rereferenced_EEG_raw = raw.copy()
    rereferenced_EEG_raw = rereferenced_EEG_raw.drop_channels(['x_dir', 'y_dir', 'z_dir'])
    #ref_channels=["F3","FZ","F4","FC5","FC1","FC2","FC6","C3","C4","CP5","CP1","CP2","CP6","P3","PZ","P4","F5","F1","F2","F6","FC3","FC4","C5","C6","CP3","CPZ","CP4","P5","P1","P2","P6"]
    if not (RerefChannels):
        print("no reref channels specified, using original ref")
    else: 
        if (RerefChannels[0] == "average"):
            print("using average reference over all electrodes")
            rereferenced_EEG_raw, ref_data = mne.set_eeg_reference(rereferenced_EEG_raw, ref_channels='average' ,copy=True)
        else:
            print("using custom electrodes for rereferencing")
            rereferenced_EEG_raw, ref_data = mne.set_eeg_reference(rereferenced_EEG_raw, ref_channels=RerefChannels ,copy=True)
    
    

    raw_EEG_Rereferenced = rereferenced_EEG_raw.copy()
    #drop channel

    #apply filter 
    Filtered_EEG_Rereferenced = raw_EEG_Rereferenced.filter(FhighpassEEG,FlowpassEEG)
    
    
    #extract events 
    PlotEvents, PlotEventDict = mne.events_from_annotations(Filtered_EEG_Rereferenced)
    PlotOnsetIndizes = np.where(PlotEvents[:,2] == OnsetNumber)[0] # S100 marker is leaving plate 
    ExcludeIndizes = np.where(PlotEvents[:,2] == ErrorNumber)[0] # S3 marker should be excluded 


    PlotOnsetIndizes = PlotOnsetIndizes[0:-1] # cut of last movement, might be after experiment
    IncludeMask = np.ones(PlotOnsetIndizes.shape)
    # Exclude S3 epochs

    #search for correct indizes without S3 errors 
    for i in range(0, len(PlotOnsetIndizes)): 
        for j in range(0, len(ExcludeIndizes)):
            if ((ExcludeIndizes[j]-1) == PlotOnsetIndizes[i] or (ExcludeIndizes[j]-2) == PlotOnsetIndizes[i]):
                IncludeMask[i] = 0


    #Epochs with an S3 error are excluded 
    PlotOnsetIndizesCorrect = PlotOnsetIndizes[IncludeMask.astype(bool)]

    #create new event matrix 
    UsedPlotEvents = np.zeros((len(PlotOnsetIndizesCorrect),3))
    UsedPlotEvents = PlotEvents[PlotOnsetIndizesCorrect,:]
     # baseline= (EpochingTimeBeforeOnset,-1)

    EEGEpochs = mne.Epochs(Filtered_EEG_Rereferenced, events = UsedPlotEvents, event_id = EventIdsUsed,tmin=t1, tmax=t2, preload=True, reject_by_annotation = True)
    if not(ChannelList): 
        EEGEpochs = mne.Epochs(Filtered_EEG_Rereferenced, events = UsedPlotEvents, event_id = EventIdsUsed,tmin=t1, tmax=t2, preload=True,reject_by_annotation = True)
    else: # is only selected channels should be epoched 
        #Filtered_EEG_RereferencedCopy = Filtered_EEG_Rereferenced.copy()
        Filtered_EEG_Rereferenced.drop_channels(ChannelList)
        EEGEpochs = mne.Epochs(Filtered_EEG_Rereferenced, events = UsedPlotEvents, event_id = EventIdsUsed,tmin=t1, tmax=t2,preload=True, reject_by_annotation = True)
    #show remaining channel names 
    RemainingEEGChannelNames = Filtered_EEG_Rereferenced.ch_names
    # give names back ! 
    
    LRPEpochObj = EEGEpochs.copy()

    #get data out as numpy array 
    LRPEpochs = EEGEpochs.get_data() 

    timeAxisEEGBatch = np.arange(t1,t2+1/FsampleEEG, step = 1/FsampleEEG) #build time axis (epoch)

    return LRPEpochs,LRPEpochObj, timeAxisEEGBatch, RemainingEEGChannelNames,Filtered_EEG_Rereferenced # shape of epochs: (epochs, channel, samples)



def CalcTestAccAndRates(PredictionLabels, TrueLabels):
    PredictionLabels = PredictionLabels.astype(int)
    TrueLabels = TrueLabels.astype(int)
    
    Negatives = np.sum(np.abs(TrueLabels-1))
    Positives = np.sum(TrueLabels)
    
    N = len(TrueLabels)
    TPs = 0 
    TNs = 0
    for index in range(0,N): 
        if (PredictionLabels[index] == 0 and TrueLabels[index] == 0):
            TNs = TNs+1
        elif (PredictionLabels[index] == 1 and TrueLabels[index] == 1): 
            TPs = TPs+1
    TNR = TNs/Negatives
    TPR = TPs/Positives
    Acc = ((TNs+TPs)/N)*100 #in percent
    return TNR, TPR, Acc
    
```

# Datasets 

```python
# Data Strings Study 
#Subject 1
DatasetStrJV43_Uni_set1 = 'data/20211216_r_RA12_intentional_unilateral_set1.vhdr'
DatasetStrJV43_Uni_set2 = 'data/20211216_r_RA12_intentional_unilateral_set2.vhdr'
DatasetStrJV43_Uni_set3 = 'data/20211216_r_RA12_intentional_unilateral_set3.vhdr'


#EEG Reading Parameter 
# Channelnumbers with EEG Data from Dataset

EEGChannelStartNumber = 0 
EEGChannelEndNumber   = 67 

# Filtering Params 
FhighpassEEG = 0.1 # in Hz 
FlowpassEEG = 4 # in Hz 


# Plotting Specs 
PlotsColor = 'black'
textSizes = 10 
Figuresizes = (8,4)

FsampleEEG = 500 #sample Frequency
Markernumber = 100

EventIdsUsed = {"MovementOnset": Markernumber} # use id 22 for epoching 
# epoche times 
EpochingTimeBeforeOnset = -1.7
t1 = EpochingTimeBeforeOnset
t2 = 0

# marker numbers for evaluation 
OnsetNumber = Markernumber
ErrorNumber = 3
```

# Load and extract EEG Data 

```python
UnilatDataList = [DatasetStrJV43_Uni_set1, DatasetStrJV43_Uni_set2, DatasetStrJV43_Uni_set3]

# append all datasets together to one giant list with raw instances 
rawList = []
for dataset in UnilatDataList: 
    raw1 = mne.io.read_raw_brainvision(dataset,preload = True, verbose = False)
    rawList.append(raw1)
    
# concetenate all raw files to one dataset 
raw = mne.concatenate_raws(rawList)


```

# Select EEG Channel

```python
ChannelList = ['FP1', 'FP2', 'F8', 'T7', 'T8', 'TP9', 'TP10', 'P7','P8', 'PO9', 'O1', 'OZ', 'O2', 'PO10', 'AF7', 'AF3', 'AF4', 'AF8', 'FT9', 'FT7', 'FT8', 'FT10', 'TP7', 'TP8', 'PO7', 'PO3', 'POZ', 'PO4', 'PO8','F7']
#ChannelList = ['F1','F2','F3','F4','F5','F6','F1','FP1', 'FP2', 'F8', 'T7', 'T8', 'TP9', 'TP10', 'P7','P8', 'PO9', 'O1', 'OZ', 'O2', 'PO10', 'AF7', 'AF3', 'AF4', 'AF8', 'FT9', 'FT7', 'FT8', 'FT10', 'TP7', 'TP8', 'PO7', 'PO3', 'POZ', 'PO4', 'PO8','F7']

RerefChannel =["average"]

LRPEpochs, LRPEpochUniObj,timeAxisEEGBatch, RemainingEEGChannelNames,Filtered_EEG_Rereferenced = RereferencingEpoching(raw, OnsetNumber, ErrorNumber,ChannelList, RerefChannel)
MeanLRPEpochs = np.mean(LRPEpochs, axis = 0) #average over epochs 

print("Shape of Epochs: ", LRPEpochs.shape)
print("Channel names: ", RemainingEEGChannelNames)

print(LRPEpochUniObj) 
```

```python
# show the data 
print(MeanLRPEpochs.shape)

plt.figure()
plt.plot(timeAxisEEGBatch, MeanLRPEpochs[4, :])
plt.title("C1-electrode")
plt.show()

# all single trials
```

```python
 # Xdawn instance
NxDAWN = 4
xd = Xdawn(n_components=NxDAWN)
    
# Fit xdawn
xd = xd.fit(LRPEpochUniObj)

LRPEpochUnixDAWN = xd.transform(LRPEpochUniObj.copy())

#LRPEpochUnixDAWN_Mean = np.mean(LRPEpochUnixDAWN, axis = 0)

#print(LRPEpochUnixDAWN.shape)
LRPEpochUniObj
```

```python
# preparing features 
print(LRPEpochUnixDAWN.shape)
ShapeData = LRPEpochUnixDAWN.shape

#features
Nsamp = 200
featuresLRP = np.zeros((Nsamp*ShapeData[0],ShapeData[1]+1))
featuresNoLRP = np.zeros((Nsamp*ShapeData[0],ShapeData[1]+1))


for i in range (0,ShapeData[1]): 
    featuresLRP[:,i] = LRPEpochUnixDAWN[:,i,-Nsamp:].reshape((Nsamp*ShapeData[0],))
    featuresNoLRP[:,i] = LRPEpochUnixDAWN[:,i,0:Nsamp].reshape((Nsamp*ShapeData[0],))


features = np.append(featuresLRP, featuresNoLRP, axis = 0)
print("Feature Shape: ", features.shape)

normalizer = Normalization(axis=-1)
normalizer.adapt(features)

normalizedfeatures = normalizer(features)
normalizedfeatures = np.array(normalizedfeatures)
normalizedfeatures[0:Nsamp*ShapeData[0],4] = 1 # encode labels 
np.random.shuffle(normalizedfeatures) # shuffle the training data 

print("Shape of features: ", normalizedfeatures.shape)

#training and validation data 
x_train = normalizedfeatures[0:int((2/3)*len(normalizedfeatures)),0:4]
y_train = normalizedfeatures[0:int((2/3)*len(normalizedfeatures)),4]
x_validation = normalizedfeatures[int((2/3)*len(normalizedfeatures)):int((4/5)*len(normalizedfeatures)),0:4]
y_validation = normalizedfeatures[int((2/3)*len(normalizedfeatures)):int((4/5)*len(normalizedfeatures)),4]
x_test = normalizedfeatures[int((4/5)*len(normalizedfeatures)):,0:4]
y_test = normalizedfeatures[int((4/5)*len(normalizedfeatures)):,4]

print("Shape train data: ", x_train.shape)
print("Shape validation data: ", x_validation.shape)
print("Shape test data: ", x_test.shape)

print("Times LRP start in s: ", timeAxisEEGBatch[-Nsamp])
print("Times NoLRP end in s: ", timeAxisEEGBatch[Nsamp])
```

```python
# MLP setup 
model = Sequential()
model.add(Dense(units=4, activation='relu', input_shape=(4,)))
model.add(Dense(units=4, activation='relu'))
model.add(Dense(units=1, activation='sigmoid')) 
```

# ML Model

```python
#show model 
print(model.summary())
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics='accuracy')
history = model.fit(x_train, 
                    y_train, 
                    epochs  = 25, 
                    batch_size= 256, 
                    validation_data=(x_validation, y_validation))
                   
```

```python
# plot the results 
history_dict = history.history
print(history_dict.keys()) 
loss_values = history_dict['loss']
val_loss_values = history_dict['val_loss']
epochs = range(1, len(loss_values)+1)
plt.figure()
plt.plot(epochs, loss_values, 'bo', label='Training loss')
plt.plot(epochs, val_loss_values, 'b', label='Validation loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()


acc_values = history_dict['accuracy']
val_acc_values = history_dict['val_accuracy']
plt.figure()
plt.plot(epochs, acc_values, 'bo', label='Training accuracy')
plt.plot(epochs, val_acc_values, 'b', label='Validation accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
```

```python
# predict Test data 
PredictionScores = model.predict(x_test)
PredictionLabels = [0 if score <0.5 else 1 for score in PredictionScores]
PredictionLabels = np.array(PredictionLabels)

TNR, TPR, Acc = CalcTestAccAndRates(PredictionLabels, y_test)

print("TNR: ",TNR)
print("TPR: ",TPR)
print("Acc: ",Acc)
```
